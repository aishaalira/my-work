# Importing Librairies

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("vgsales.csv")
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
platforms = df ['Platform'].value_counts()
platforms.shape
```

```python
plt.figure(figsize=(10,6))
plt.title("Used Platform")
color = ["blue","red","brown","green","cyan","indigo"]
plt.bar(platforms.index, platforms.values, color=color)
plt.xlabel("Name of Platform")
plt.ylabel("Values")
plt.xticks(rotation =90)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.title("Used Platforms")
plt.pie(platforms.values, labels=platforms.index, autopct= "%1.1f%%", shadow=True)
plt.show()
```

```python
sorted_data = df.head(10)
platforms = sorted_data['Platform'].value_counts()
sorted_data.shape
```

```python
plt.figure(figsize=(10,6))
plt.title("Used Platforms")
values = []
explode = []
for x in platforms.values:
    values.append(float(x))
max_value = max(values)
for x in values:
    if x == max_value:
           explode.append(0.1)
    else:
        explode.append(0)
plt.pie(platforms.values, labels=platforms.index, explode=explode, autopct= "%1.1f%%", shadow=True)
plt.show()
```

```python
ordered_data = df.sort_values(by="Global_Sales", ascending=False)
ordered_data.head(10)
```

```python
names = sorted_data['Name']
sales = sorted_data['Global_Sales']
```

```python
plt.figure(figsize=(10,6))
plt.title("Games with the highest Global Sales")
colors =["green","red","cyan","indigo","black"]
plt.bar(names, sales, color = colors)
plt.xlabel("Names of Games")
plt.ylabel("Global Sales")
plt.xticks(rotation = 70)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.title("Games with the highest Global Sales")
sns.barplot(x=names, y=sales)
plt.xlabel("Names of Games")
plt.ylabel("Global Sales")
plt.xticks(rotation = 70)
plt.show()
```
